```python

# Contact Book using Dictionary

contacts = {
    "Rahul": "9876543210",
    "Sneha": "9876543211",
    "Amit": "9876543212"
}

print("CONTACT BOOK")
print("----------------")

for name, phone in contacts.items():
    print("Name:", name, "| Phone:", phone)

print("----------------")
print("Total Contacts:", len(contacts))
```

```python

# Simple Calculator using Functions

def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b

# Given numbers
a = 20
b = 10

print("First Number:", a)
print("Second Number:", b)
print("Addition:", add(a, b))
print("Subtraction:", subtract(a, b))
print("Multiplication:", multiply(a, b))
print("Division:", divide(a, b))
```

```python

# Weather App using OpenWeather API

import requests

city = "Pune"
api_key = "YOUR_API_KEY"

url = (
    "https://api.openweathermap.org/data/2.5/weather"
    f"?q={city}&appid={api_key}&units=metric"
)

print("WEATHER APP")
print("--------------------")
print("City:", city)

if api_key == "YOUR_API_KEY":
    print("Please add your OpenWeather API key to get live weather.")
else:
    response = requests.get(url, timeout=10)

    if response.status_code == 200:
        data = response.json()
        print("Temperature:", data["main"]["temp"], "°C")
        print("Weather:", data["weather"][0]["description"])
        print("Humidity:", data["main"]["humidity"], "%")
    else:
        print("Could not get weather data.")
        print("Status code:", response.status_code)
```

```python

# Live Cryptocurrency Prices

import requests

url = "https://api.coingecko.com/api/v3/simple/price"

params = {
    "ids": "bitcoin,ethereum,dogecoin",
    "vs_currencies": "usd"
}

print("LIVE CRYPTOCURRENCY PRICES")
print("--------------------------")

try:
    response = requests.get(url, params=params, timeout=15)
    response.raise_for_status()
    data = response.json()

    print("Bitcoin (BTC): $", data["bitcoin"]["usd"])
    print("Ethereum (ETH): $", data["ethereum"]["usd"])
    print("Dogecoin (DOGE): $", data["dogecoin"]["usd"])

except requests.exceptions.RequestException:
    print("Unable to fetch live prices. Please try again later.")
```
