```python
# Create a sample text file
with open("sample.txt", "w") as file:
    file.write("Hello Python\n")
    file.write("Welcome to Week 3\n")
    file.write("File Handling Practice\n")

# Read the file and count lines
with open("sample.txt", "r") as file:
    lines = file.readlines()

print("Total lines:", len(lines))
```

```python
# Create first text file
with open("file1.txt", "w") as f:
    f.write("Hello Python\n")

# Create second text file
with open("file2.txt", "w") as f:
    f.write("Welcome to File Handling\n")

# Merge both files into one file
with open("file1.txt", "r") as f1:
    data1 = f1.read()

with open("file2.txt", "r") as f2:
    data2 = f2.read()

with open("merged.txt", "w") as f:
    f.write(data1 + data2)

# Display merged content
with open("merged.txt", "r") as f:
    print(f.read())
```

```python
import pandas as pd

# Create sample data
data = {
    "Name": ["Amit", "Sneha", "Rahul", "Priya"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 90, 78, 95]
}

# Convert data into a DataFrame
df = pd.DataFrame(data)

# Save data as a CSV file
df.to_csv("students.csv", index=False)

# Load the CSV file
df = pd.read_csv("students.csv")

# Display and analyze data
print("CSV Data:")
print(df)

print("\nTotal Students:", len(df))
print("Average Marks:", df["Marks"].mean())
print("Highest Marks:", df["Marks"].max())
```
